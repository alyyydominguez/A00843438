# Análisis de ventas de una tienda

## Pregunta de análisis
¿Qué producto tiene las mayores ventas?

## Origen de los datos
Se utilizarán datos inventados para fines educativos. No se utilizará información real de clientes ni de empresas.

## Resultados esperados
Se espera comparar las ventas de diferentes productos, identificar cuál tiene las mayores ventas y representar los resultados mediante una gráfica sencilla.

```python

import sys
import pandas as pd
import matplotlib

print("Versión de Python:", sys.version)
print("Versión de pandas:", pd.__version__)
print("Versión de matplotlib:", matplotlib.__version__)
```

```python

import pandas as pd

# Base de datos con ventas ficticias
datos = {
    "Producto": ["Camisetas", "Pantalones", "Tenis", "Sudaderas", "Gorras"],
    "Unidades_vendidas": [120, 85, 150, 65, 95],
    "Precio": [250, 500, 1200, 600, 180]
}

df = pd.DataFrame(datos)

# Mostrar la tabla
df
```

```python

# Calculo de las ventas totales de cada producto
df["Ventas_totales"] = df["Unidades_vendidas"] * df["Precio"]

# Identificar el producto que generó más dinero
producto_top = df.loc[df["Ventas_totales"].idxmax()]

print("Producto con mayores ventas:", producto_top["Producto"])
print("Ingresos generados: $", producto_top["Ventas_totales"])

# Mostrar la tabla actualizda
df
```

```python

import matplotlib.pyplot as plt

# Gráfica de barras
plt.figure(figsize=(9, 5))
plt.bar(df["Producto"], df["Ventas_totales"], color="skyblue")

# Título y nombres a los ejes
plt.title("Ventas totales por producto")
plt.xlabel("Producto")
plt.ylabel("Ingresos ($ MXN)")

# Mostrar la gráfica
plt.show()
```

## Conclusión

Al analizar los datos, encontre que los tenis son el producto que genera mayores ingresos, con un total de $180,000 MXN. Esto se debe a que tienen un precio más alto y también registran una cantidad importante de ventas.
